```python
# ==========================================
# MODULE 02: DATA ANALYSIS WITH PYTHON
# ==========================================

# 1. IMPORT LIBRARIES
import numpy as np
import pandas as pd

print("Libraries imported successfully!")

# 2. NUMPY BASICS (Working with arrays)
print("\n--- NUMPY ARRAYS ---")
arr_1d = np.array([10, 20, 30, 40, 50])
print("1D Array:", arr_1d)
print("Mean value:", np.mean(arr_1d))
print("Max value:", np.max(arr_1d))

# 3. IMPORTING & CREATING A SAMPLE CSV DATASET
print("\n--- IMPORTING & CREATING DATASET ---")
# Creating a sample dataset dictionary with some missing values (NaN) to practice cleaning
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR', None],
    'Age': [25, 30, np.nan, 28, 35, 40],
    'Salary': [50000, 45000, 60000, np.nan, 55000, 70000]
}

df = pd.DataFrame(data)
print("Original DataFrame with Missing Values:")
print(df)

# Save to CSV and reload to practice CSV importing
df.to_csv('employee_data.csv', index=False)
df_loaded = pd.read_csv('employee_data.csv')
print("\nSuccessfully imported CSV dataset!")

# 4. EXPLORING DATA
print("\n--- DATA EXPLORATION ---")
print("First 3 rows:")
print(df_loaded.head(3))
print("\nDataFrame Info:")
print(df_loaded.info())
print("\nStatistical Summary:")
print(df_loaded.describe())

# 5. HANDLING MISSING VALUES & DATA CLEANING
print("\n--- HANDLING MISSING VALUES ---")
print("Check missing values per column:")
print(df_loaded.isnull().sum())

# Fill missing Ages with the median age
median_age = df_loaded['Age'].median()
df_loaded['Age'] = df_loaded['Age'].fillna(median_age)

# Fill missing Department with 'Unknown'
df_loaded['Department'] = df_loaded['Department'].fillna('Unknown')

# Fill missing Salary with mean salary
mean_salary = df_loaded['Salary'].mean()
df_loaded['Salary'] = df_loaded['Salary'].fillna(mean_salary)

print("\nDataFrame after Cleaning & Handling Missing Values:")
print(df_loaded)

# 6. BASIC DATA ANALYSIS TECHNIQUES
print("\n--- BASIC DATA ANALYSIS ---")
print("Average Salary after cleaning:", df_loaded['Salary'].mean())
print("\nGroup by Department and find average salary:")
print(df_loaded.groupby('Department')['Salary'].mean())
```
